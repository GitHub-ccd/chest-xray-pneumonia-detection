This notebook modernizes the legacy TensorFlow/Keras pipeline into a state-of-the-art **PyTorch** deep learning system featuring:

3. **Explainable AI (XAI)**: **Grad-CAM (Gradient-weighted Class Activation Mapping)** overlays to highlight pulmonary opacity attention zones.

## 2. Exploratory Data Analysis & Albumentations Augmentations

```python
# Load datasets
train_ds = PneumoniaDataset("data", split="train", transform=get_train_transforms(224))
val_ds   = PneumoniaDataset("data", split="val", transform=get_valid_transforms(224))
test_ds  = PneumoniaDataset("data", split="test", transform=get_valid_transforms(224))

# Class distribution
train_labels = [label for _, label in train_ds.samples]
normal_cnt = train_labels.count(0)
pneu_cnt = train_labels.count(1)

plt.figure(figsize=(6, 4))
plt.bar(["NORMAL", "PNEUMONIA"], [normal_cnt, pneu_cnt], color=["#2ecc71", "#e74c3c"])
plt.title("Train Set Imbalance Distribution")
plt.ylabel("Image Count")
for i, cnt in enumerate([normal_cnt, pneu_cnt]):
    plt.text(i, cnt + 50, str(cnt), ha='center', fontweight='bold')
plt.show()
```
